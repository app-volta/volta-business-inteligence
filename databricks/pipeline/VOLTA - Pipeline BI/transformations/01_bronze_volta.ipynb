```python
from pyspark import pipelines as dp


@dp.materialized_view(
    name="workspace.bronze.company_raw",
    comment="Recorte da empresa necessário aos indicadores de BI."
)
def company_raw():
    return spark.read.table(
        "volta_postgres.public.company"
    ).select("id", "name")


@dp.materialized_view(
    name="workspace.bronze.area_raw",
    comment="Recorte das áreas necessário aos indicadores de BI."
)
def area_raw():
    return spark.read.table(
        "volta_postgres.public.area"
    ).select("id", "company_id", "sector_name", "location_description")


@dp.materialized_view(
    name="workspace.bronze.waste_type_raw",
    comment="Categorias de resíduos usadas nos indicadores."
)
def waste_type_raw():
    return spark.read.table(
        "volta_postgres.public.waste_type"
    ).select("id", "category", "default_risk_level")


@dp.materialized_view(
    name="workspace.bronze.incident_raw",
    comment="Ocorrências de origem, com os campos usados pelo dashboard."
)
def incident_raw():
    return spark.read.table(
        "volta_postgres.public.incident"
    ).select(
        "id",
        "company_id",
        "area_id",
        "waste_type_id",
        "registered_at",
        "estimated_quantity",
        "priority",
        "status",
        "contamination_level",
    )


@dp.materialized_view(
    name="workspace.bronze.collection_raw",
    comment="Coletas de origem para os KPIs de operação."
)
def collection_raw():
    return spark.read.table(
        "volta_postgres.public.collection"
    ).select(
        "id",
        "incident_id",
        "cooperative_id",
        "requested_at",
        "scheduled_at",
        "current_status",
        "collection_type",
        "urgent",
    )


@dp.materialized_view(
    name="workspace.bronze.collection_status_raw",
    comment="Histórico de status usado para calcular tempo de resolução."
)
def collection_status_raw():
    return spark.read.table(
        "volta_postgres.public.collection_status"
    ).select("collection_id", "status", "changed_at")


@dp.materialized_view(
    name="workspace.bronze.cooperative_raw",
    comment="Nome e coordenadas das cooperativas associadas às coletas."
)
def cooperative_raw():
    return spark.read.table(
        "volta_postgres.public.cooperative"
    ).select("id", "name", "latitude", "longitude", "average_rating")
```
